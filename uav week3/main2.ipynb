```python
import cv2
import time
from ultralytics import YOLO

# Load YOLOv8 Small model
model = YOLO("yolov8s.pt")

# Open webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Webcam not found!")
    exit()

while True:

    success, frame = cap.read()

    if not success:
        break

    # Start time
    start = time.time()

    # YOLO Prediction
    results = model.predict(frame, conf=0.25)

    # Draw Bounding Boxes
    annotated_frame = results[0].plot()

    # End time
    end = time.time()

    # Calculate FPS
    fps = 1 / (end - start)

    # Display FPS
    cv2.putText(
        annotated_frame,
        f"FPS: {fps:.2f}",
        (20,40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0,255,0),
        2
    )

    # Show Output
    cv2.imshow("YOLOv8s Webcam Detection", annotated_frame)

    # Press q to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
