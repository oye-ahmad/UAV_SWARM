```python
import os
import random
import shutil
from pathlib import Path
```

```python
# Original dataset

IMAGE_DIR = r"C:\Users\MUHAMMADAHMAD\Desktop\VisDrone2019-DET-train\VisDrone2019-DET-train\images"
ANNOTATION_DIR = r"C:\Users\MUHAMMADAHMAD\Desktop\VisDrone2019-DET-train\VisDrone2019-DET-train\annotations"

# Output dataset

OUTPUT_DIR = r"C:\Users\MUHAMMADAHMAD\Desktop\VizDrone_Dataset_Prepared"
```

```python
TRAIN_RATIO = 0.70
VALID_RATIO = 0.20
TEST_RATIO = 0.10
```

```python
folders = [

    "train/images",
    "train/annotations",

    "valid/images",
    "valid/annotations",

    "test/images",
    "test/annotations"

]

for folder in folders:
    os.makedirs(os.path.join(OUTPUT_DIR, folder), exist_ok=True)

print("Folders Created Successfully")
```

```python
image_files = []

for file in os.listdir(IMAGE_DIR):

    if file.endswith(".jpg"):

        image_files.append(file)

print("Total Images :", len(image_files))
```

```python
dataset = []

missing_annotations = []

for image_name in image_files:

    annotation_name = image_name.replace(".jpg", ".txt")

    image_path = os.path.join(IMAGE_DIR, image_name)

    annotation_path = os.path.join(ANNOTATION_DIR, annotation_name)

    if os.path.exists(annotation_path):

        dataset.append((image_path, annotation_path))

    else:

        missing_annotations.append(image_name)
```

```python
print("Valid Image-Annotation Pairs :", len(dataset))

print("Missing Annotation Files :", len(missing_annotations))

if len(missing_annotations) > 0:

    print()

    print("Images without annotations:")

    for file in missing_annotations:

        print(file)
```

```python
random.seed(42)

random.shuffle(dataset)

print("Dataset Shuffled Successfully")
```

```python
total = len(dataset)

train_size = int(total * TRAIN_RATIO)

valid_size = int(total * VALID_RATIO)

test_size = total - train_size - valid_size

print("Total :", total)

print("Train :", train_size)

print("Validation :", valid_size)

print("Test :", test_size)
```

```python
train_data = dataset[:train_size]

valid_data = dataset[train_size:train_size+valid_size]

test_data = dataset[train_size+valid_size:]
```

```python
def copy_dataset(data, split_name):

    image_output = os.path.join(OUTPUT_DIR, split_name, "images")

    annotation_output = os.path.join(OUTPUT_DIR, split_name, "annotations")

    for image_path, annotation_path in data:

        shutil.copy(image_path, image_output)

        shutil.copy(annotation_path, annotation_output)

    print(f"{split_name} set copied successfully.")
```

```python
copy_dataset(train_data, "train")
```

```python
copy_dataset(valid_data, "valid")
```

```python
copy_dataset(test_data, "test")
```

```python
for split in ["train","valid","test"]:

    image_count = len(os.listdir(os.path.join(OUTPUT_DIR, split, "images")))

    annotation_count = len(os.listdir(os.path.join(OUTPUT_DIR, split, "annotations")))

    print()

    print(split.upper())

    print("-----------------------")

    print("Images :", image_count)

    print("Annotations :", annotation_count)
```

```python
for split in ["train","valid","test"]:

    image_dir = os.path.join(OUTPUT_DIR, split, "images")

    annotation_dir = os.path.join(OUTPUT_DIR, split, "annotations")

    images = sorted(os.listdir(image_dir))

    annotations = sorted(os.listdir(annotation_dir))

    errors = 0

    for image in images:

        annotation = image.replace(".jpg", ".txt")

        if annotation not in annotations:

            errors += 1

    print(f"{split.upper()} Pair Errors :", errors)
```

```python
for root, dirs, files in os.walk(OUTPUT_DIR):

    level = root.replace(OUTPUT_DIR, "").count(os.sep)

    indent = " " * 4 * level

    print(f"{indent}{os.path.basename(root)}/")

    subindent = " " * 4 * (level + 1)

    for file in files[:5]:      # Display first five files only

        print(f"{subindent}{file}")

    if len(files) > 5:

        print(f"{subindent}...")
```
