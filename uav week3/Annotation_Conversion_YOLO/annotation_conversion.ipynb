```python
import os
import cv2
from pathlib import Path
```

```python
DATASET_PATH = r"C:\Users\MUHAMMADAHMAD\Desktop\VizDrone_Dataset_Prepared"
```

```python
CLASS_MAPPING = {

    1:0,
    2:1,
    3:2,
    4:3,
    5:4,
    6:5,
    7:6,
    8:7,
    9:8,
    10:9

}
```

```python
splits = ["train","valid","test"]

for split in splits:

    label_dir = os.path.join(DATASET_PATH,
                             split,
                             "labels")

    os.makedirs(label_dir,
                exist_ok=True)

print("Label folders created.")
```

```python
def convert_annotation(annotation_path, image_path, output_path):

    image = cv2.imread(image_path)

    if image is None:
        print(f"Could not read image: {image_path}")
        return

    img_height, img_width = image.shape[:2]

    with open(annotation_path, "r") as file:
        lines = file.readlines()

    yolo_annotations = []

    for line in lines:

        values = line.strip().split(",")

        if len(values) != 8:
            continue

        x = float(values[0])
        y = float(values[1])
        w = float(values[2])
        h = float(values[3])

        score = int(values[4])
        class_id = int(values[5])

        # Ignore invalid regions
        if score == 0:
            continue

        # Ignore class 11 ("Others")
        if class_id not in CLASS_MAPPING:
            continue

        x_center = x + w / 2
        y_center = y + h / 2

        x_center /= img_width
        y_center /= img_height
        w /= img_width
        h /= img_height

        class_id = CLASS_MAPPING[class_id]

        yolo_annotations.append(
            f"{class_id} "
            f"{x_center:.6f} "
            f"{y_center:.6f} "
            f"{w:.6f} "
            f"{h:.6f}"
        )

    # Write output ONCE after processing all objects
    with open(output_path, "w") as file:
        file.write("\n".join(yolo_annotations))
```

```python
total_images = 0

for split in splits:

    image_dir = os.path.join(DATASET_PATH,
                             split,
                             "images")

    annotation_dir = os.path.join(DATASET_PATH,
                                  split,
                                  "annotations")

    label_dir = os.path.join(DATASET_PATH,
                             split,
                             "labels")

    images = sorted(os.listdir(image_dir))

    for image_name in images:

        image_path = os.path.join(image_dir,
                                  image_name)

        annotation_name = image_name.replace(".jpg",
                                             ".txt")

        annotation_path = os.path.join(annotation_dir,
                                       annotation_name)

        output_path = os.path.join(label_dir,
                                   annotation_name)

        convert_annotation(annotation_path,
                           image_path,
                           output_path)

        total_images += 1

print()

print("Conversion Complete")

print()

print("Total Images Converted :", total_images)
```
