```python
from ultralytics import YOLO
import cv2
import time
import os
import numpy as np
```

```python
pt_model = YOLO(r"C:\Users\MUHAMMADAHMAD\Desktop\InternShip Week 4\uav week4\uav week4\runs\detect\train-12\weights\best.pt")

onnx_model = YOLO(r"C:\Users\MUHAMMADAHMAD\Desktop\InternShip Week 4\uav week4\uav week4\runs\detect\train-12\weights\best.onnx")
```

```python
test_folder = r"C:\Users\MUHAMMADAHMAD\Desktop\InternShip Week 4\uav week4\uav week4\VizDrone_Dataset_Prepared\test\images"

image_files = []

for file in os.listdir(test_folder):

    if file.endswith(".jpg"):

        image_files.append(
            os.path.join(test_folder, file)
        )

print("Total Test Images:", len(image_files))
```

```python
pt_total_time = 0

pt_total_detections = 0

for image in image_files:

    start = time.time()

    results = pt_model.predict(
        source=image,
        verbose=False
    )

    end = time.time()

    pt_total_time += end-start

    pt_total_detections += len(results[0].boxes)

print("PT Model - Total Time:", pt_total_time)
print("PT Model - Total Detections:", pt_total_detections)
```

```python
onnx_total_time = 0

onnx_total_detections = 0

for image in image_files:

    start = time.time()

    results = onnx_model.predict(
        source=image,
        verbose=False
    )

    end = time.time()

    onnx_total_time += end-start

    onnx_total_detections += len(results[0].boxes)
```

```python
num_images = len(image_files)

pt_avg_time = pt_total_time / num_images

onnx_avg_time = onnx_total_time / num_images

pt_fps = 1 / pt_avg_time

onnx_fps = 1 / onnx_avg_time

print("PT Model - Average Time per Image:", pt_avg_time)
print("PT Model - FPS:", pt_fps)

print("ONNX Model - Average Time per Image:", onnx_avg_time)
print("ONNX Model - FPS:", onnx_fps)
```

```python
onnx_metrics = onnx_model.val(
    data=r"C:\Users\MUHAMMADAHMAD\Desktop\InternShip Week 4\uav week4\uav week4\VizDrone_Dataset_Prepared\data.yaml",
    verbose=False
)
```

```python
print("Precision :", onnx_metrics.box.mp)
print("Recall    :", onnx_metrics.box.mr)
print("mAP50     :", onnx_metrics.box.map50)
print("mAP50-95  :", onnx_metrics.box.map)
```
